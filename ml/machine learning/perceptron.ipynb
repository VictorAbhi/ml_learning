```python
import numpy as np
import matplotlib.pyplot as plt

# Define inputs and weights
X = np.array([1.5, 3.0, 3.5])  # Inputs: X1=1.5, X2=3, X3=3.5
W = np.array([0.25, 0.11, 0.05])  # Weights: W1=0.25, W2=0.11, W3=0.05
bias = 0.25  # Initial bias

# Sigmoid activation function
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# Compute weighted sum and output
def compute_output(X, W, bias):
    z = np.dot(X, W) + bias  # Weighted sum: Z = X1*W1 + X2*W2 + X3*W3 + bias
    y = sigmoid(z)  # Apply sigmoid activation
    return y

# Calculate initial output
y_initial = compute_output(X, W, bias)
print(f"Initial Output (bias=0.25): Y = {y_initial:.4f}")

# Vary bias and compute outputs
bias_values = [0.1, 0.5, 1.0]
outputs = []
for b in bias_values:
    y = compute_output(X, W, b)
    outputs.append(y)
    print(f"Output for bias={b}: Y = {y:.4f}")

# Plot Y vs. bias
plt.plot(bias_values, outputs, marker='o', linestyle='-', color='blue')
plt.xlabel('Bias')
plt.ylabel('Output (Y)')
plt.title('Effect of Bias on Perceptron Output')
plt.grid(True)
plt.show()
```

### Explanation of Bias Effect
The bias term shifts the weighted sum Z, directly affecting the sigmoid output. A higher bias increases Z, pushing the output Y closer to 1 due to the sigmoid function’s shape (Sigmoid(Z) = 1 / (1 + e^(-Z))). For example, as bias increased from 0.1 to 1.0, Z rose from 1.0 to 1.9, causing Y to increase from 0.7271 to 0.8676. This demonstrates that the bias adjusts the perceptron’s activation threshold, controlling how easily the neuron produces a high output.
